```python
import pandas as pd
import numpy as np

df_vendas = pd.read_csv("../dados_publicos/vendas_brutas.csv")
df_custo = pd.read_csv("../dados_publicos/custos_brutos.csv")
df_ads = pd.read_csv("../dados_publicos/ads_brutos.csv")


print("Vendas:", df_vendas.shape)
print("Custos:", df_custo.shape)
print("Ads:", df_ads.shape)

```

```python

df_analise = df_vendas[[
    "ID do pedido",
    "Data de criação do pedido",
    "Status do pedido",
    "Status da Devolução / Reembolso",
    "Hora do pagamento do pedido",
    "Nome do Produto",
    "Quantidade",
    "Returned quantity",
    "Preço original",
    "Preço acordado",
    "Subtotal do produto",
    "Valor Total",
    "Desconto do vendedor",
    "Cupom do vendedor",
    "Desconto Shopee da Leve Mais por Menos",
    "Taxa de transação",
    "Taxa de comissão bruta",
    "Taxa de comissão líquida",
    "Taxa de serviço bruta",
    "Taxa de serviço líquida",
    "Taxa de envio pagas pelo comprador",
    "Taxa de Envio Reversa",
    "Total global",
    "Valor estimado do frete",
    "UF",
    "Cidade.1"
    ]]

df_analise["Status da Devolução / Reembolso"] = df_analise["Status da Devolução / Reembolso"].fillna("Sem Devolução")
df_analise = df_analise.drop_duplicates(subset=["ID do pedido"])
df_analise["Data de criação do pedido"] = pd.to_datetime(df_analise["Data de criação do pedido"], format= "%Y-%m-%d %H:%M")

display(df_analise)
display(df_analise.info())
display(df_analise [["Valor Total", "Preço original"]])




```

```python

#ANALISE VENDAS  


Faturamento = df_analise ["Valor Total"].sum()
Pedidos = df_analise ["ID do pedido"].nunique()
Qte_Pedidos = df_analise ["Quantidade"].sum()
Tiket_Medio = Faturamento / Pedidos

df_analise ["Mes"] = df_analise ["Data de criação do pedido"].dt.to_period("M")
Vendas_por_mes = df_analise.groupby("Mes")["ID do pedido"].nunique()
Maior_mes_de_venda = Vendas_por_mes.idxmax()
Maior_num_vendas = Vendas_por_mes.max()

Taxas_totais = (df_analise["Taxa de transação"].sum() 
+ df_analise["Taxa de comissão líquida"].sum() 
+ df_analise["Taxa de serviço líquida"].sum())

Valor_apos_taxa = Faturamento - Taxas_totais
Percentual_apos_taxa = (Valor_apos_taxa / Faturamento)*100


print ("===RESUMO DE VENDAS===")
print(f"Faturamento: R${Faturamento:.2f}")
print(f"Total de pedidos: {Pedidos}")
print(f"Qte Vendida: {Qte_Pedidos}")
print(f"Ticket Médio: R${Tiket_Medio:.2f}")

display(df_analise ["Status do pedido"].value_counts())
display(df_analise ["Status da Devolução / Reembolso"].value_counts())

print(f"Periodo com mais vendas:{Maior_mes_de_venda}")
print(Maior_num_vendas)

display(df_analise[[ "Taxa de transação", 
                  "Taxa de comissão bruta",
                  "Taxa de comissão líquida",
                  "Taxa de serviço bruta",
                  "Taxa de serviço líquida",
                  "Taxa de Envio Reversa"]].sum())

print(f"Totais pagos em taxas: R${Taxas_totais:.2f}")
print(f"Valor após taxas: R${Valor_apos_taxa:.2f}")
print(f"Percentual após taxas: {Percentual_apos_taxa:.2f}%")


```

```python
dfanalise_ads = df_ads [["periodo",
                         "Impressões",
                          "Cliques", 
                          "Conversões", 
                          "Itens Vendidos",
                          "GMV", "Despesas",
                          "ROAS",
                          "ACOS"]]

corregir_colunas = ["Impressões",
                          "Cliques", 
                          "Conversões", 
                          "Itens Vendidos",
                          "GMV", "Despesas",
                          "ROAS",
                          "ACOS"]

dfanalise_ads[corregir_colunas] = (dfanalise_ads[corregir_colunas]/100)
investimento_total = dfanalise_ads["Despesas"].sum()
gmv_total = dfanalise_ads["GMV"].sum()
roas_geral = gmv_total / investimento_total

melhor_periodo = dfanalise_ads.loc[dfanalise_ads["ROAS"].idxmax()]
pior_periodo = dfanalise_ads.loc[dfanalise_ads["ROAS"].idxmin()]

acos_geral = (investimento_total / gmv_total) * 100

melhor_acos = dfanalise_ads.loc[dfanalise_ads["ACOS"].idxmin()]
pior_acos = dfanalise_ads.loc[dfanalise_ads["ACOS"].idxmax()]

display(dfanalise_ads)
display(dfanalise_ads.info())
display(dfanalise_ads.describe())

display(f"Investimento Total em ADS: R${investimento_total:.2f}")
display(f"GMV Total dos ADS: R${gmv_total:.2f}")
display(f"ROAS Geral: {roas_geral:.2f}")
display(f"ACOS Geral: {acos_geral:.2f}%")
display(dfanalise_ads[["periodo", "Despesas", "GMV", "ROAS"]])

print("MELHOR PERÍODO")
print (f"Período: {melhor_periodo['periodo']}")
print (f"ROAS: {melhor_periodo['ROAS']:.2f}")
print (f"GMV: {melhor_periodo['GMV']:.2f}")
print (f"Despesas: {melhor_periodo['Despesas']:.2f}")

print("\nPIOR PERÍODO")
print (f"Período: {pior_periodo['periodo']}")
print (f"ROAS: {pior_periodo['ROAS']:.2f}")
print (f"GMV: {pior_periodo['GMV']:.2f}")
print (f"Despesas: {pior_periodo['Despesas']:.2f}")

display(dfanalise_ads[["periodo", "Despesas", "GMV", "ROAS", "ACOS"]])

print("MELHOR ACOS")
print(f"Período: {melhor_acos['periodo']}")
print(f"ACOS: {melhor_acos['ACOS']:.2f}%")

print("\nPIOR ACOS")
print(f"Período: {pior_acos['periodo']}")
print(f"ACOS: {pior_acos['ACOS']:.2f}%")
```

```python
custo_anel = 18.00
custo_embalagem = 6.00
custo_unitario = custo_anel  + custo_embalagem
quantidade_vendida = df_analise["Quantidade"].sum()
custo_total_produtos = quantidade_vendida * custo_unitario

taxa_comissao = df_analise["Taxa de comissão líquida"].sum()
taxa_servico = df_analise["Taxa de serviço líquida"].sum()
taxa_transacao = df_analise["Taxa de transação"].sum()

taxas_totais1 = taxa_comissao + taxa_servico + taxa_transacao

custos_total = (custo_total_produtos + taxas_totais1 + investimento_total)

lucro_estimado = Faturamento - custos_total



display(df_custo)
display(df_custo.info())
display(df_custo.describe())

print(f"Custo Unitário: R$ {custo_unitario:.2f}")
print(f"Quantidade Total Vendida: {quantidade_vendida}")
print(f"Custo Total dos Produtos: {custo_total_produtos:.2f}")
print(f"Total de Taxas da Shopee: R$ {taxas_totais1:.2f}")
print(f"Custos Totais: R$ {custos_total:.2f}")

print(f"Faturmento Total: R$ {Faturamento:.2f}")
print(f"Lucro Estimado: R$ {lucro_estimado:.2f}")

display(f"Faturamento: R$ {Faturamento:.2f}")

print(df_analise.columns.tolist())

display(df_analise[["Subtotal do produto", "Valor Total", "Taxa de comissão líquida", "Taxa de serviço líquida", "Taxa de transação", "Total global"]].head(10))
display(df_analise["Status do pedido"].value_counts())
```

```python
df_vendas_validas = df_analise[df_analise["Status do pedido"]!= "Cancelado"].copy()
quantidade_venda_valida = df_vendas_validas["Quantidade"].sum()
faturamento_valido = df_vendas_validas["Valor Total"].sum()

custo_total_produtos = quantidade_venda_valida * custo_unitario

taxa_comissao_valida = df_vendas_validas["Taxa de comissão líquida"].sum()
taxa_servico_valida = df_vendas_validas["Taxa de serviço líquida"].sum()
taxa_transacao_valida = df_vendas_validas["Taxa de transação"].sum()

taxa_totais_validas = (taxa_comissao_valida + taxa_servico_valida + taxa_transacao_valida)

custos_totais_corrigidos = (custo_total_produtos + taxa_totais_validas + investimento_total)
resultado_operacao = faturamento_valido - custos_totais_corrigidos

print(f"Registros válidos: {len(df_vendas_validas)}")
print(f"Quantidade venda válida: {quantidade_venda_valida}")
print(f"Faturamento válido: R$ {faturamento_valido:.2f}")
print(f"Custo totais dos produtos: R$ {custo_total_produtos:.2f}")
print(f"Taxas Válidas das Shopee: R$ {taxa_totais_validas:.2f}")
print(f"Custos Totais: R$ {custos_totais_corrigidos:.2f}")
print(f"Resultado da operação: R$ {resultado_operacao:.2f}")
```

```python
resultado_sem_ads = (faturamento_valido - custo_total_produtos - taxa_totais_validas)
margem_sem_ads = (resultado_sem_ads / faturamento_valido) * 100
margem_com_ads = (resultado_operacao / faturamento_valido) * 100

roas_equilibrio = 1 / (margem_sem_ads / 100)


print(f"Resultado sem ADS: R$ {resultado_sem_ads:.2f}")
print(f"Margem sem Ads: {margem_sem_ads:.2f}%")
print(f"Margem com Ads: {margem_com_ads:.2f}%")
print(f"ROAS equilíbrio: {roas_equilibrio:.2f}")
```
